#📁 1. os – Operating System Interface

```python
import os

# Get current working directory
print("Current Directory:", os.getcwd())

# List files and directories in current directory
print("Files and Folders:", os.listdir())

# Create a new folder
os.mkdir("demo_folder")  # Creates folder named 'demo_folder'

# Rename a folder
os.rename("demo_folder", "renamed_folder")

# Remove a folder (make sure it's empty)
os.rmdir("renamed_folder")
```

#📆 2. datetime – Date and Time Handling

```python
from datetime import datetime, date, timedelta

# Current date and time
now = datetime.now()
print("Now:", now)

# Only today's date
today = date.today()
print("Today's Date:", today)

# Create a specific date
dob = date(2000, 1, 1)
print("DOB:", dob)

# Date arithmetic
yesterday = today - timedelta(days=1)
tomorrow = today + timedelta(days=1)

print("Yesterday:", yesterday)
print("Tomorrow:", tomorrow)

# Format date to string
formatted = now.strftime("%d-%m-%Y %H:%M:%S")
print("Formatted DateTime:", formatted)
```

#🎲 3. random – Random Numbers & Choices

```python
import random

# Generate random integer between 1 and 100
print("Random Int:", random.randint(1, 100))

# Generate random float between 0 and 1
print("Random Float:", random.random())

# Choose a random item from a list
colors = ['red', 'green', 'blue']
print("Random Color:", random.choice(colors))

# Shuffle a list
random.shuffle(colors)
print("Shuffled List:", colors)

# Generate a random sample of multiple items
print("Sample:", random.sample(range(1, 50), 5))
```

#📦 4. collections – Specialized Containers
🔹 A. Counter

```python
from collections import Counter

# Count occurrences of elements in a list
fruits = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
count = Counter(fruits)
print("Fruit Count:", count)
```

🔹 B. defaultdict

```python
from collections import defaultdict

# Returns a default value for missing keys
marks = defaultdict(lambda: 0)
marks['Math'] = 90
marks['Science'] = 80
print("English Marks (not set):", marks['English'])  # Returns 0
```

🔹 C. namedtuple

```python
from collections import namedtuple

# Create a Point object with x and y
Point = namedtuple('Point', 'x y')
p1 = Point(10, 20)
print("Point:", p1)
print("x:", p1.x, "y:", p1.y)
```

🔹 D. deque

```python
from collections import deque

# Double-ended queue
dq = deque([1, 2, 3])
dq.append(4)        # Add to right
dq.appendleft(0)    # Add to left
dq.pop()            # Remove from right
dq.popleft()        # Remove from left
print("Deque after operations:", dq)
```

## Assignment: Event Log Generator and Analyzer

**Objective:** Create a script that simulates generating event logs, saves them to a file, and then analyzes the events using the `os`, `datetime`, `random`, and `collections` libraries.

**Task Breakdown:**

1.  **Setup (`os`):**
    *   Create a new directory named `event_logs` to store your generated log files.

2.  **Log Generation (`datetime`, `random`):**
    *   Generate a list of 50 random events. Each event should have:
        *   A random timestamp within the last 7 days (`datetime.now()`, `timedelta`).
        *   A random event type from a predefined list (e.g., `'LOGIN'`, `'LOGOUT'`, `'PURCHASE'`, `'ERROR'`, `'VIEW'`) (`random.choice`).
    *   Write these events to a text file named `daily_events_YYYY-MM-DD.log` inside the `event_logs` directory, where `YYYY-MM-DD` corresponds to the date of the event's timestamp. Format each log entry as: `[TIMESTAMP] - [EVENT_TYPE]`.

3.  **Log Analysis (`collections`, `os`):**
    *   Read the generated log files from the `event_logs` directory.
    *   Use `collections.Counter` to count the occurrences of each `EVENT_TYPE` across all log files.
    *   Use `collections.defaultdict` to group events by their date (e.g., to see how many events occurred on each day).
    *   Print the total count of each event type and the number of events per day.

4.  **Cleanup (Optional, `os`):**
    *   After analysis, you can optionally remove the `event_logs` directory and its contents (`os.remove`, `os.rmdir`).

**Hints:**
*   Remember to open files in write mode (`'w'`) to create them and read mode (`'r'`) to read them.
*   Pay attention to string formatting for dates and log entries.

```python
import os

os.makedirs("event_logs", exist_ok=True)

print("event_logs folder created")
```

```python
import os
import random
print("enter logs is created")
event_types=['LOGIN','LOGOUT','PURCHASE','ERROR','VIEW']
res=random.choices(event_types,k=1)
print(res)
events=[]
num_events=5
for i in range(num_events):
  days_ago=random.randint(0,6)
  hours_ago=random.randint(0,23)
  minutes_ago=random.randint(0,59)
  seconds_ago=random.randint(0,59)
  timestamp=datetime.now()-timedelta(days=days_ago,hours=hours_ago,
                                     minutes=minutes_ago,seconds=seconds_ago)
  events.append({'timestamp':timestamp,'type':event_types})
daily_events={}
for event in events:
  date_str=event['timestamp'].strftime('%Y-%m-%d')
  if date_str not in daily_events:
    daily_events[date_str]=[]
  daily_events[date_str].append(event)
for date_str,event_list in daily_events.items():
    file=os.path.join('event_logs',f'daily_events_{date_str}.log')
    with open(file,'w') as f:
      for event in event_list:
        f.write(f"[{event['timestamp'].strftime('%Y-%m-%d %H:%M:%S')}]-{event['type']}\n")
        print(f"Generated log file: {file} with {len(event_list)} events.")

print(f"Successfully generated {num_events} events across {len(daily_events)} log files.")



```

```python
import os
from collections import Counter, defaultdict
log_eg=os.listdir('event_logs')
print(log_eg)
event_count = Counter()
daily_count = defaultdict(int)
for filename in os.listdir("event_logs"):
    filepath = os.path.join("event_logs", filename)
    with open(filepath, "r") as f:
        for line in f:
            parts = line.strip().split(" - ")
            if len(parts) > 1:
                event_type = parts[1]
                event_count[event_type] += 1
print("Event Type Counts:")
for event_type, count in event_count.items():
    print(f"  {event_type}: {count}")
```
