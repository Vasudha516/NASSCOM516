```python
'''Write a Python program to find the second largest unique element
in a list without using the
sort() function.
Example: Input: [12,45,23,45,67,12] Output: 45'''
# Input list
numbers = [12, 45, 23, 45, 67, 12]

largest = None
second_largest = None

for num in numbers:
    if largest is None or num > largest:
        if num != largest:
            second_largest = largest
            largest = num
    elif num != largest and (second_largest is None or num > second_largest):
        second_largest = num

print("Second largest unique element:", second_largest)
```

```python
'''6. Longest Consecutive Sequence
Given an unsorted list, find the length of the longest consecutive sequence.
Example: [100,4,200,1,3,2] -> 4'''
Eg=[100,4,200,1,3,2]
Eg.sort()
print(Eg)
longest = 1
current = 1
for i in range(1, len(Eg)):
    if Eg[i] == Eg[i-1] + 1:
        current += 1
    else:
        current = 1

    if current > longest:
        longest = current

print("Longest consecutive sequence length:", longest)

```

```python
'''7. Matrix Operations using Nested Lists
Read two matrices using nested lists and perform matrix addition, subtraction, and
multiplication.'''
row1=int(input("enter row"))
col1=int(input("enter column"))
matrix1=[[int(input()) for i in range(col1)]for j in range(row1)]
print(matrix1)
row2=int(input("enter row"))
col2=int(input("enter column"))
matrix2=[[int(input()) for i in range(col1)]for j in range(row1)]
print(matrix2)
print("addition of matrix")
add_matrix=[[0 for i in range(col1)]for j in range(row1)]
for i in range(row1):
  for j in range(col2):
    add_matrix[i][j]=matrix1[i][j]+matrix2[i][j]
print(add_matrix)
print("subtraction of matrix")
sub_matrix=[[0 for i in range(col1)]for j in range(row1)]
for i in range(row1):
  for j in range(col2):
    sub_matrix[i][j]=matrix1[i][j]-matrix2[i][j]
print(sub_matrix)
print("multiplication of matrix")
mul_matrix=[[0 for i in range(col1)]for j in range(row1)]
for i in range(row1):
  for j in range(col1):
    sum=0
    for k in range(row2):
      sum+=matrix1[i][k]*matrix2[k][j]
    mul_matrix[i][j]=sum
print(mul_matrix)
```

```python
'''8. Spiral Traversal
Given an N×N matrix stored as nested lists, print the elements in spiral order.'''
matrix1 = []
row = int(input("enter row"))
col = int(input("enter column"))
for i in range(row):
    a = []
    for j in range(col):
        a.append(int(input()))
    matrix1.append(a)
print("Original matrix:")
print(matrix1)
matrix = [[0 for j in range(col)] for i in range(row)]
top = 0
bottom = len(matrix1) - 1
left = 0
right = len(matrix1[0]) - 1
r = 0
c = 0
while top <= bottom and left <= right:
    for i in range(left, right + 1):
        matrix[r][c] = matrix1[top][i]
        c += 1
        if c == col:
            c = 0
            r += 1
    top += 1
    for i in range(top, bottom + 1):
        matrix[r][c] = matrix1[i][right]
        c += 1
        if c == col:
            c = 0
            r += 1
    right -= 1
    if top <= bottom:
        for i in range(right, left - 1, -1):
            matrix[r][c] = matrix1[bottom][i]
            c += 1
            if c == col:
                c = 0
                r += 1

        bottom -= 1
    if left <= right:
        for i in range(bottom, top - 1, -1):
            matrix[r][c] = matrix1[i][left]
            c += 1
            if c == col:
                c = 0
                r += 1
        left += 1
print("Spiral matrix:")
print(matrix)
```

```python
'''9. Majority Element
Determine whether a list contains a majority element (appearing more than n/2 times).
Print the element or 'No Majority Element'.'''
list_ex=[2,2,1,2,2,3]
n=len(list_ex)/2
freq={}
for i in range(len(list_ex)):
  if(list_ex[i] in freq):
    freq[list_ex[i]]+=1
  else:
    freq[list_ex[i]]=1
print(freq)
majority=0
for i in freq:
  if freq[i]>n:
    majority=i
if majority==0:
  print("no majority element")
else:
  print("majority element",majority)


```

```python
'''
10. Stock Buy and Sell
Given daily stock prices, determine the maximum profit possible by buying once and selling
once.
'''
stock_prices=[2,3,7,1,5,7,6,8]
buy=stock_prices[0]
profit=0
for i in range(len(stock_prices)):
  if stock_prices[i]<buy:
    buy=stock_prices[i]
  else:
    current_profit =stock_prices[i] - buy
    if current_profit > profit:
      profit = current_profit

print("Maximum Profit:", profit)


```
