## 6. Pivot Tables & Merge/Join

```python
# -----------------------------------------------------------
# 🔹 6A. DEMO — pivot_table and merge
# -----------------------------------------------------------
pivot = df.pivot_table(
    values='survived', index='class', columns='sex',
    aggfunc='mean', observed=True
)
print('Survival rate pivot (class x sex):')
print(pivot)

left = pd.DataFrame({'id': [1, 2, 3], 'name': ['A', 'B', 'C']})
right = pd.DataFrame({'id': [1, 2, 4], 'score': [90, 85, 70]})
merged = left.merge(right, on='id', how='left')
print('\nLeft join example:')
print(merged)
```

## How to Parse Dates from a Column

To convert a column containing date strings into datetime objects, you can use the `pd.to_datetime()` function. This is crucial for performing time-series analysis or any operations that require date-specific functionalities.

```python
# Create a sample date column in the 'df' DataFrame for demonstration
df['sample_date_str'] = pd.Series([
    '2023-01-01',
    '2023-01-02',
    '2023-01-03',
    '2023-01-04',
    '2023-01-05'
] * (len(df) // 5 + 1)).head(len(df))

print("Original 'sample_date_str' column type:")
print(df['sample_date_str'].dtype)

# Convert the 'sample_date_str' column to datetime objects
df['parsed_date'] = pd.to_datetime(df['sample_date_str'])

print("\nConverted 'parsed_date' column type:")
print(df['parsed_date'].dtype)

print("\nFirst 5 rows of the new 'parsed_date' column:")
display(df[['sample_date_str', 'parsed_date']].head())
```
